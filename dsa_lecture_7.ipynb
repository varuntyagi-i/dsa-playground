```python
'''

Binary Search

search_space = [1,3,5,6,7,13,24,26,34,45,55,65]

'''

def search(target, left, right):
    
    while left <= right:
    
        mid = (left+right) // 2
    
        if search_space[mid] == target:
            return mid

        elif search_space[mid]<target:
            left = mid+1

        else:
            right = mid-1

    return -1



search_space = [1,3,5,6,7,13,24,26,34,45,55,65]
print(search(7, 0, len(search_space)-1))
```

```python
'''

Find the Floor of a number

given_number = [1,3,5,6,7,13,24,26,34,45,55,65]

eg
ip: 4 , op: 3
ip: 29 , op: 26
ip: 34 , op: 34

'''

given_number = [1,3,5,6,7,13,24,26,34,45,55,65]

floor_chk = 29

def get_floor(left, right):

    ans = -1

    while left <= right:

        mid = (left+right) // 2

        if given_number[mid] == floor_chk:
            return given_number[mid]
       
        elif given_number[mid] > floor_chk:
            right = mid-1

        else:
            ans = given_number[mid]
            left = mid+1

    return ans

output_response = get_floor(0, len(given_number)-1)
print(output_response)
```
