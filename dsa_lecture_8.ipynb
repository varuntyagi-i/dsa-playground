```python
'''

Binary Search

search_space = [1,3,5,6,7,13,24,26,34,45,55,65]

'''

def search(target, left, right):
    
    while left <= right:
    
        mid = (left+right) // 2
    
        if search_space[mid] == target:
            return mid

        elif search_space[mid]<target:
            left = mid+1

        else:
            right = mid-1

    return -1



search_space = [1,3,5,6,7,13,24,26,34,45,55,65]
print(search(7, 0, len(search_space)-1))
```

```python
'''

Find the Floor of a number

given_number = [1,3,5,6,7,13,24,26,34,45,55,65]

eg
ip: 4 , op: 3
ip: 29 , op: 26
ip: 34 , op: 34

'''

given_number = [1,3,5,6,7,13,24,26,34,45,55,65]

floor_chk = 29

def get_floor(left, right):

    ans = -1

    while left <= right:

        mid = (left+right) // 2

        if given_number[mid] == floor_chk:
            return given_number[mid]
       
        elif given_number[mid] > floor_chk:
            right = mid-1

        else:
            ans = given_number[mid]
            left = mid+1

    return ans

output_response = get_floor(0, len(given_number)-1)
print(output_response)
```

```python
'''
Google and Fractal

N = [] <contains distinct element>

Find local minima in the area : a number smaller than available neighbour (just one number is enough)

A[i-1] > A[i] < A[i+1]

eg.
ip : 3,6,1,0,9,15,8
op : 3,0,8

ip : 5, 8, 12, 3
op : 5, 8, 12, 3

Solve in O(logn)
'''

A = [6, 1, 0, 9, 15, 18]
N = len(A)

def get_local_minima():
    
    if A[0] < A[1]:
        return A[0]
    
    if A[N-1] < A[N-2]:
        return A[N-1]
    
    left = 1
    right = len(A)-2
    
    while left <= right:
        mid = (left+right) // 2
        if A[mid-1] < A[mid] < A[mid+1]:
            right = mid - 1
        elif A[mid-1] > A[mid] > A[mid+1]:
            left = mid + 1
        else:
            return A[mid]

value = get_local_minima()
print(value)
```
