```python
''' 

Every element occurs twice except for 1 value. Find the unique value in array.
Note: Duplicate values are adjacent to each other.

eg:
ip: [3,3,1,1,8,8,10,10,9,6,6,2,2,4,4]
op: 9

----

Hint:

left of unique element the duplicate pair start with EVEN index
right of unique element the duplicate pair start with ODD index


----

'''

A = [3,3,1,1,8,8,10,10,9,6,6,2,2,4,4]

def unique():
    
    left = 0
    right = len(A)-1
    
    if len(A)==1:
        return A[0]
    
    if A[0] != A[1]:
        return A[0]
    
    if A[right] != A[right-1]:
        return A[right]
    
    while left <= right:
        
        mid = (left+right) // 2

        if A[mid-1] != A[mid] and A[mid+1] != A[mid]:
            return A[mid]
        
        if A[mid-1] == A[mid]:
            mid = mid - 1

        if mid % 2 == 0:
            left = mid + 2
        
        else:
            right = right - 1



print(unique())
```

```python
''' 

Given a number, get the SQRT of that number. Provide the FLOOR value where required.

eg
sqrt(25) = 5
sqrt(20) = 4
sqrt(10) = 3

'''

def sqrt(number):
    l = 1
    r = number
    ans = number

    while l <= r:
        
        mid = (l+r) // 2
        
        if mid*mid > number:
            r = mid-1
        
        else:
            ans = mid
            l = mid+1 
        
        print(l, r, mid, ans)

    return ans

print(sqrt(1000))
```
