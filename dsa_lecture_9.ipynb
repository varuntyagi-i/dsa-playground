```python
''' 

Every element occurs twice except for 1 value. Find the unique value in array.
Note: Duplicate values are adjacent to each other.

eg:
ip: [3,3,1,1,8,8,10,10,9,6,6,2,2,4,4]
op: 9

----

Hint:

XOR property

 101
+101
=000

 000
+101
=101

----

'''

A = [3,3,1,1,8,8,10,10,9,6,6,2,2,4,4]

def unique():
    
    left = 0
    right = len(A)-1
    
    if len(A)==1:
        return A[0]
    
    if A[0] != A[1]:
        return A[0]
    
    if A[right] != A[right-1]:
        return A[right]
    
    while left <= right:
        
        mid = (left+right) // 2

        if A[mid-1] != A[mid] and A[mid+1] != A[mid]:
            return A[mid]
        
        if A[mid-1] == A[mid]:
            mid = mid - 1

        if mid % 2 == 0:
            left = mid + 2
        
        else:
            right = right - 1



print(unique())
```
