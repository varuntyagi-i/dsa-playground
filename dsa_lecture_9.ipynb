```python
''' 

Every element occurs twice except for 1 value. Find the unique value in array.
Note: Duplicate values are adjacent to each other.

eg:
ip: [3,3,1,1,8,8,10,10,9,6,6,2,2,4,4]
op: 9

----

Hint:

left of unique element the duplicate pair start with EVEN index
right of unique element the duplicate pair start with ODD index


----

'''

A = [3,3,1,1,8,8,10,10,9,6,6,2,2,4,4]

def unique():
    
    left = 0
    right = len(A)-1
    
    if len(A)==1:
        return A[0]
    
    if A[0] != A[1]:
        return A[0]
    
    if A[right] != A[right-1]:
        return A[right]
    
    while left <= right:
        
        mid = (left+right) // 2

        if A[mid-1] != A[mid] and A[mid+1] != A[mid]:
            return A[mid]
        
        if A[mid-1] == A[mid]:
            mid = mid - 1

        if mid % 2 == 0:
            left = mid + 2
        
        else:
            right = right - 1



print(unique())
```

```python
''' 

Given a number, get the SQRT of that number. Provide the FLOOR value where required.

eg
sqrt(25) = 5
sqrt(20) = 4
sqrt(10) = 3

'''

def sqrt(number):
    l = 1
    r = number
    ans = number

    while l <= r:
        
        mid = (l+r) // 2
        
        if mid*mid > number:
            r = mid-1
        
        else:
            ans = mid
            l = mid+1 
        
        # print(l, r, mid, ans)

    return ans

print(sqrt(1000))
```

```python
''' 

Information:

Given N array element : max subarray sum of length k

A = [-3, 4, -2, 5, 3, -2, 8, 2, -1, 4]

sliding window =>   time complexity : O(n)
                    space complexity : O(1)

Question:

Google

Given an array +ve integers, find MAX "K" such that max subarray sum of len "K" <= B.
B is an integers

Eg:
input                    
A = [3, 2, 5, 4, 6, 3, 7, 2]
B = 20

Solution
K   max_sum   answer
1       7       1
2       10      2
3       16      3
4       20      4
5       25      4 


'''
A = [3, 2, 5, 4, 6, 3, 7, 2]
B = 10

def submax(len_arr_k):
    arr_sum = sum(A[:len_arr_k])
    tmp_sum = arr_sum
    for i in range(len_arr_k, len(A)):
        tmp_sum -= A[i-len_arr_k]
        tmp_sum += A[i]
        if tmp_sum > arr_sum:
            arr_sum = tmp_sum
    return tmp_sum

def get_max_k():
    l=0
    h=len(A)
    ans = 0
    while l <= h:
        m = (l+h)//2
        sum = submax(m)

        if sum < B:
            ans = m
            l = m + 1
        else:
            h = m - 1
    
    return ans

print(get_max_k())
```

```python
''' 

Count pair : "a""g" 
such that  
i < j 
i.e. s[i]='a' and s[j]='g' 

eg
ip: "baagdcag" 
op: 
1,3 
1,7 
2,3 
2,7
6,7

'''

s = "baagdcag"
def get_pair_count():
    g_count = 0
    ans = 0
    for i in range(len(s)-1, 0, -1):
        if s[i] == "g":
            g_count += 1
        elif s[i] == "a":
            ans += g_count

    return ans

print(get_pair_count()) 
```
