```python
'''
= Recursion ==
count all subsets of a set of size N
N=3
{},{1},{2},{3},{1,2},{1,3},{2,3},{1,2,3}
'''

def count_subsets(n):
    if n==0:
        return 1
    return 2*count_subsets(n-1)

n=int(input())
count_subsets(n)
```

```python
''' 
cs(4)
    -> cs(2)
            -> cs(1)
                -> cs(0)
                    prod = return 1

                -> condition : n=0
                    return 1*1

            -> condition : n=1
                    return 2*1*1

    -> condition : n=2
        n%2:
            2*2
            return 4

condition : n=4
    4*4
    return 16

'''
''' 
Thinking Process :

I know count_subsets(n) = 2^n

Can I reduce n?
Yes, split n into n//2 and n//2.

If n is even:
    n = k + k
    so 2^n = 2^k * 2^k

If n is odd:
    n = k + k + 1
    so 2^n = 2 * 2^k * 2^k
'''

def count_subsets(n):
    if n==0:
        return 1

    prod = count_subsets(n//2)

    if n%2==0:
        return prod*prod
    else:
        return 2*prod*prod

n=int(input())
count_subsets(n)
```

```python
''' 
== Recursion ==

Fibonnaci Series

Solution:
f(n) = f(n-1) + f(n-2)
Complexity = O(2^n)

Question:
Find solution and reduce the complexity to O(n)
0<=input<=n

Solution:
0,1,1,2,3,5,8,13,21,....

'''
def f(n, a, b):
    if n==0:
        return a
    return f(n-1, b, a+b)
    
def fib(n):
    return f(n, 0, 1)

n=int(input())
fib(n)
```
