```python
'''
= Recursion ==
count all subsets of a set of size N
N=3
{},{1},{2},{3},{1,2},{1,3},{2,3},{1,2,3}
'''

def count_subsets(n):
    if n==0:
        return 1
    return 2*count_subsets(n-1)

n=int(input())
count_subsets(n)
```

```python
''' 
cs(4)
    -> cs(2)
            -> cs(1)
                -> cs(0)
                    prod = return 1

                -> condition : n=0
                    return 1*1

            -> condition : n=1
                    return 2*1*1

    -> condition : n=2
        n%2:
            2*2
            return 4

condition : n=4
    4*4
    return 16

'''
''' 
Thinking Process :

I know count_subsets(n) = 2^n

Can I reduce n?
Yes, split n into n//2 and n//2.

If n is even:
    n = k + k
    so 2^n = 2^k * 2^k

If n is odd:
    n = k + k + 1
    so 2^n = 2 * 2^k * 2^k
'''

def count_subsets(n):
    if n==0:
        return 1

    prod = count_subsets(n//2)

    if n%2==0:
        return prod*prod
    else:
        return 2*prod*prod

n=int(input())
count_subsets(n)
```

```python
''' 
== Recursion ==

Fibonnaci Series

Solution:
f(n) = f(n-1) + f(n-2)
Complexity = O(2^n)

Question:
Find solution and reduce the complexity to O(n)
0<=input<=n

Solution:
0,1,1,2,3,5,8,13,21,....

'''
def f(n, a, b):
    if n==0:
        return a
    return f(n-1, b, a+b)
    
def fib(n):
    return f(n, 0, 1)

n=int(input())
fib(n)
```

```python
''' 
== Recusion ==

nCk = n! / ( k! * (n-k)! )

C(n,n) = C(n,0) = 1
C(n,k) = C(n, n-k)
C(n,k) = C(n-1, k-1) + C(n-1, k)


Eg.
n=5 (possible value : 1,2,3,4,5)
k=2

Output : [ [1,2] [1,3] [1,4] [1,5] [2,3] [2,4] [2,5] [3,4] [3,5] [4,5] ]

'''

def find_comb(n, k):

    c: list[list[int]] = []
    s: list[int] = []

    def f(i:int) -> None:
        if len(s) == k:
            c.append(s[:])
            return
        
        if i==n+1:
            return
        
        else:
            s.append(i)
            f(i+1)
            s.pop()
            f(i+1)
    
    f(1)
    return c

input_arr = list(map(int, list(input())))
find_comb(input_arr[0], input_arr[1])
```

```python
''' 
Rope Length (Max)

Eg.
1. n:5, a=2, b=5, c=1
    Answer: 5 ([1,1,1,1,1])
2. n:5, a=2, b=5, c=1
    Answer: -1 (no such combination present)
3. n:23, a=12, b=9, c=11
    answer: 2  ((11, 12) or (12, 11))
'''
def rope(n, a, b, c):
    if n==0:
        return 0
    if n <= -1:
        return -1
    
    result =  max(  
                rope(n-a, a, b, c),
                rope(n-b, a, b, c),
                rope(n-c, a, b, c)
                )
    
    if result == -1:
        return -1
    
    return result+1


input_arr = list(map(int, list(input())))
find_comb(input_arr[0], input_arr[1], input_arr[2], input_arr[3])
```
